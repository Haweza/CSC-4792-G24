```python
import pdfplumber
import pandas as pd

columns = {
    "Strategies": (0, 106),
    "Programmes": (106, 184),
    "Sub Program": (184, 354),
    "Location": (354, 517),
    "YR 1": (517, 550),
    "YR 2": (550, 591),
    "YR 3": (591, 627),
    "YR 4": (627, 663),
    "YR 5": (663, 694),
    "Responsible": (694, 800)
}

rows = []

with pdfplumber.open("IDP_2024_2034.pdf") as pdf:
    page = pdf.pages[155]
    words = page.extract_words()

    for top in sorted(set(round(w["top"], 1) for w in words)):
        row_words = [w for w in words if round(w["top"], 1) == top]

        row = {}

        for column, (xmin, xmax) in columns.items():
            selected = [
                w["text"]
                for w in row_words
                if xmin <= w["x0"] < xmax
            ]

            row[column] = " ".join(selected)

        if any(row.values()):
            rows.append(row)

df_test = pd.DataFrame(rows)

print(df_test.to_string(index=False))
```

```python
import pytesseract
from pytesseract import Output
import pandas as pd
admin_data = []

for page_num in range(1, 6):
    data = pytesseract.image_to_data(
        pages[page_num],
        output_type=Output.DATAFRAME
    )
    
    data = data.dropna(subset=["text"])
    data = data[data["text"].str.strip() != ""]
    
    data["page"] = page_num + 1
    
    admin_data.append(data)

admin_data = pd.concat(admin_data, ignore_index=True)

print("Rows:", len(admin_data))
```

```python
admin_data.head(30)
```

```python
page2_table = admin_data[
    (admin_data["page"] == 2) &
    (admin_data["top"] >= 1300)
]

page2_table[[
    "left", "top", "width", "height", "conf", "text"
]].head(50)
```

```python
administrative_dataset = []

for page, page_df in admin_data.groupby("page"):

    # Sort OCR elements from top to bottom
    page_df = page_df.sort_values("top")

    # Group words into rows based on vertical position
    rows = []
    current_row = []
    current_y = None

    for _, item in page_df.iterrows():

        if current_y is None:
            current_y = item["top"]
            current_row = [item]

        elif abs(item["top"] - current_y) <= 30:
            current_row.append(item)

        else:
            rows.append(current_row)
            current_row = [item]
            current_y = item["top"]

    if current_row:
        rows.append(current_row)

    # Convert each OCR row into columns
    for row in rows:

        row = sorted(row, key=lambda x: x["left"])

        no = []
        name = []
        position = []
        ward = []

        for item in row:

            x = item["left"]
            text = str(item["text"]).strip()

            if x < 440:
                no.append(text)

            elif x < 950:
                name.append(text)

            elif x < 1490:
                position.append(text)

            else:
                ward.append(text)

        administrative_dataset.append({
            "No": " ".join(no),
            "Name": " ".join(name),
            "Position": " ".join(position),
            "Ward": " ".join(ward)
        })
```

```python
admin_df = pd.DataFrame(administrative_dataset)
```
